```python
from pymongo import MongoClient
import pandas as pd
```

```python
client = MongoClient("mongodb://localhost:27017/")
db = client["taller4_db"]
collection = db["raw_data"]
data = list(collection.find())
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df = pd.DataFrame({
    "nombre": df["name"].apply(lambda x: x.get("common")),
    "poblacion": df["population"],
    "region": df["region"],
    "area": df["area"],
    "capital": df["capital"]
})

df.head()
```
