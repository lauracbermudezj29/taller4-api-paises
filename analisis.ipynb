```python
from pymongo import MongoClient
import pandas as pd
```

```python
client = MongoClient("mongodb://localhost:27017/")
db = client["taller4_db"]
collection = db["raw_data"]
data = list(collection.find())
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df = pd.DataFrame({
    "nombre": df["name"].apply(lambda x: x.get("common")),
    "poblacion": df["population"],
    "region": df["region"],
    "area": df["area"],
    "capital": df["capital"]
})

df.head()
```

```python
df.loc[df["area"].idxmax()] #Pais con mayor area
```

```python
df["region"].value_counts() #Cantidad de paises por region
```

```python
df["area"].mean() #Promedio del area de los paises
```

```python
import matplotlib.pyplot as plt

df["region"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Distribución de países por región")
plt.show()
```

```python
df["region"].value_counts().plot(
    kind="bar"
)

plt.title("Cantidad de países por región")
plt.xlabel("Región")
plt.ylabel("Cantidad")
plt.show()
```

```python
df["poblacion"].plot(kind="hist", bins=20)
plt.xscale("log")
plt.title("Distribución de población (escala log)")
plt.xlabel("Población")
plt.show()
```
